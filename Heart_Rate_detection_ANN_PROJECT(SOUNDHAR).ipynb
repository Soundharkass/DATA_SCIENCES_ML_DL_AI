```python
import pandas as pd
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

import warnings
# Ignore all warnings
warnings.filterwarnings("ignore")
%matplotlib inline
```

```python
# Load dataset
data = pd.read_csv("Sleep_health_and_lifestyle_dataset.csv")
data

```

```python
# Remove extra spaces
data.columns = data.columns.str.strip()
```

```python
# Select needed columns
X = data[['Sleep Duration','Heart Rate', 'Physical Activity Level','Daily Steps']].values
```

```python
def convert_mood(heart_rate_percent):
    if 50 <= heart_rate_percent <= 60:
        return 0   # Zone 1: Very light activity
    elif 61 <= heart_rate_percent <= 70:
        return 1   # Zone 2: Light activity
    elif 71 <= heart_rate_percent <= 80:
        return 2   # Zone 3: Moderate exercise
    elif 81 <= heart_rate_percent <= 90:
        return 3   # Zone 4: Hard exercise
    elif 91 <= heart_rate_percent <= 100:
        return 4   # Zone 5: Maximum effort
    else:
        return -1  # Outside zones


y = data['Heart Rate'].apply(convert_mood).values
```

```python
# Create ANN model
model = Sequential()
model.add(Dense(8, input_dim=4, activation='relu'))
model.add(Dense(4, activation='softmax'))
```

```python
# Compile model
model.compile(loss='sparse_categorical_crossentropy',
              optimizer='adam',
              metrics=['accuracy'])
```

```python
model.fit(X, y, epochs=50)
```

```python
# Take user input
sleep = float(input("Enter Sleep Hours: "))
heart_rate_percent = float(input("Enter Heart rate Level (40-85): "))
activity = float(input("Enter Physical Activity Level (30-100) : "))
daily_steps =float(input("Enter the Daily Steps per day: "))
```

```python
# Convert into numpy array
test = np.array([[sleep,heart_rate_percent,activity,daily_steps]])
pred = model.predict(test)
labels = ["Zone 1: Very light activity", "Zone 2: Light activity", "Zone 3: Moderate exercise","Zone 4: Hard exercise","# Zone 5: Maximum effort"]
print("Mood:", labels[np.argmax(pred)])
```
