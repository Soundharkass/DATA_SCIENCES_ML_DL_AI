## FEATURE SELECTIOIN 

### 1. **Filter Numeric Features**

> * KNN works on distance metrics, so categorical features need encoding. First, separate numeric features:

```python
numeric_features = df.select_dtypes(include=['int64','float64']).columns
print("Numeric Features:", numeric_features)
```

### 2. **Encode Categorical Features**
> * Convert categorical variables (like `fuelType`, `transmission`, `model`) into numeric form:

```python
from sklearn.preprocessing import LabelEncoder
for col in ['fuelType','transmission','model']:
    if col in df.columns:
        df[col] = LabelEncoder().fit_transform(df[col])
```

### 3. **Correlation Analysis**
> * Check which features correlate strongly with the target (e.g., `price`):

```python
corr = df.corr()
print(corr['price'].sort_values(ascending=False))
```

### 4. **Feature Importance via SelectKBest**
Use statistical tests to pick top features:

```python
from sklearn.feature_selection import SelectKBest, f_regression

X = df.drop("price", axis=1)
y = df["price"]

selector = SelectKBest(score_func=f_regression, k=5)
X_new = selector.fit_transform(X, y)

selected_features = X.columns[selector.get_support()]
print("Selected Features:", selected_features)
```

### 5. **Recursive Feature Elimination (Optional)**
For classification tasks:

```python
from sklearn.feature_selection import RFE
from sklearn.linear_model import LogisticRegression

X = df.drop("price", axis=1)   # features
y = df["price"]                # target

log_reg = LogisticRegression(max_iter=1000)
rfe = RFE(log_reg, n_features_to_select=5)
rfe.fit(X, y)

print("RFE Selected Features:", X.columns[rfe.support_])
```

## 🔑 Feature Selection Workflow
## ✅ Insights
- Likely important features: **year, mileage, engine size, transmission, fuel type**.  
- Less useful features: **model name** (too many categories, may add noise).  
- Target variable: **price** (for regression) or categorical column (for classification).  

## MODEL CREATIONS

# 1.Scale features

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
print(X_scaled)
```

### 2. Train/Test Split

```python
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)
```

### 4. Build KNN Model

```python
# Initialize KNN with default k=5
knn = KNeighborsClassifier(n_neighbors=5)

# Train model
knn.fit(X_train, y_train)

# Predictions
y_pred = knn.predict(X_test)
print(y_pred)
```

## 📌 Model Creation:

1. **Objective**  
   - Build a K‑Nearest Neighbors (KNN) model to classify BMW car prices into categories (Low, Medium, High).  
   - Focus on a simple pipeline without hyperparameter tuning.

2. **Data Preparation**  
   - Encoded categorical features (`fuelType`, `transmission`, `model`) using Label Encoding.  
   - Scaled numerical features with `StandardScaler` to ensure fair distance calculations.  
   - Target variable (`price`) was discretized into three categories for classification.

3. **Train/Test Split**  
   - Dataset split into **80% training** and **20% testing** using `train_test_split`.  
   - Ensures unbiased evaluation of model performance.

4. **Model Building**  
   - Used `KNeighborsClassifier` with **k=5** (default).  
   - Trained the model on the scaled training data.

5. **Prediction & Evaluation**  
   - Predictions made on the test set.  
   - Evaluated using:
     - **Accuracy** → overall correctness of predictions.  
     - **Confusion Matrix** → class‑wise prediction errors.  
     - **Classification Report** → precision, recall, F1‑score for each price category.

6. **Insights**  
   - KNN achieved good accuracy, with most misclassifications occurring between **Medium** and **High** price ranges.  
   - Features like **year, mileage, engine size, transmission, and fuel type** were most influential.  
   - The model demonstrates how distance‑based learning can classify cars into meaningful price categories.

## MODEL EVALUATION

### 1. Accuracy

```python
print("Accuracy:", accuracy_score(y_test, y_pred))
```

- **Definition**: Ratio of correctly predicted samples to total samples.  
- **Use**: Quick overall performance measure.  
- **Limitation**: Can be misleading if classes are imbalanced.

### 2. Confusion Matrix

```python
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))
```

- **Definition**: Table showing correct vs. incorrect predictions across all classes.  
- **Use**: Helps identify which categories (Low, Medium, High price) are misclassified.  
- **Insight**: Diagonal values = correct predictions; off-diagonal = misclassifications.

### 3. Classification Report

```python
print("Classification Report:\n", classification_report(y_test, y_pred))
```

 **Metrics included**:
  - **Precision**: Out of predicted positives, how many are correct.  
  - **Recall**: Out of actual positives, how many were correctly predicted.  
  - **F1-Score**: Harmonic mean of precision and recall (balances both).  
  - **Support**: Number of samples per class.  


## 📊 Model Evaluation

After training the KNN model, we evaluate its performance on the test set. This ensures the model generalizes well and isn’t just memorizing the training data.
## 📌 Insights
- **Accuracy** gives a quick snapshot of performance.  
- **Confusion Matrix** shows where the model struggles (e.g., Medium vs High price overlap).  
- **Precision/Recall/F1** provide a deeper look at class-specific performance.  
- For imbalanced datasets, **F1-score** is more reliable than accuracy.
