````python
import cv2
import time
from ultralytics import YOLO

# 1. Load YOLOv8 model (pre-trained on COCO dataset)
# COCO dataset includes 'car' as a class
model = YOLO("yolov8n.pt")  # you can use yolov8s.pt or yolov8m.pt for better accuracy

# 2. Open video source (0 = webcam, or replace with "traffic.mp4")
video_path = 0  # change to "traffic.mp4" if you want to use a file
cap = cv2.VideoCapture(video_path)

# Optional: Stop automatically after a fixed duration (e.g., 60 seconds)
max_duration = 60  # seconds
start_time = time.time()

# 3. Loop through video frames
while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    # 4. Run YOLO detection
    results = model(frame)

    # 5. Draw detections on frame
    annotated_frame = results[0].plot()  # YOLO built-in plotting

    # 6. Show video
    cv2.imshow("Car Detection", annotated_frame)

    # Press 'q' to quit OR stop after max_duration
    if cv2.waitKey(1) & 0xFF == ord('q'):
        print("Stopped manually (q pressed).")
        break
    if time.time() - start_time > max_duration:
        print("Stopped automatically after 60 seconds.")
        break

# 7. Release resources
cap.release()
cv2.destroyAllWindows()
````

````python
### 🎯 Filter Only Cars
If you want **only cars**:
```python
for result in results:
    for box in result.boxes:
        cls_id = int(box.cls[0])
        label = model.names[cls_id]
        if label == "car":  # only cars
            x1, y1, x2, y2 = box.xyxy[0]
            cv2.rectangle(frame, (int(x1), int(y1)), (int(x2), int(y2)), (0,255,0), 2)
            cv2.putText(frame, label, (int(x1), int(y1)-10),
                        cv2.FONT_HERSHEY_SIMPLEX, 0.9, (0,255,0), 2)
````
