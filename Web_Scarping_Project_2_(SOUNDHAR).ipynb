```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
url="https://www.amazon.in/Python-Web-Scraping-Data-Collection/dp/B0F9G4PCHG"
```

```python
response=requests.get(url)
```

```python
Soup = BeautifulSoup(response.text,"html.parser")
```

```python
table = Soup.find("table")
rows =table.find_all("tr")
```

```python
# Check if table exists before trying to find rows
table = Soup.find("table")
if table:  # Check if table was found (not None)
    rows = table.find_all("tr")
else:
    print("No table found in the HTML")
    rows = []  # Set rows to empty list if no table found

# Alternative approach - you can also use try-except
# try:
#     table = Soup.find("table")
#     rows = table.find_all("tr")
# except AttributeError:
#     print("No table found or table has no rows")
#     rows = []
```

```python
data =[]

for row in rows:
  cols=row.find_all(["td","th"])
  cols=[c.text.strip() for c in cols]
  data.append(cols)
```

```python
df = pd.DataFrame(data[1:],columns = data[0])
df
```

```python
df.to_csv("dataset.csv",index=False)
print("CSV CREATED SUCESSFULLY")
print(df)
```
